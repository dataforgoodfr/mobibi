```python
# /// script
# dependencies = [
#     "pandas",
#     "gspread",
#     "google-auth",
# ]
# ///

import gspread
import pandas as pd

# === 1. Vos valeurs configurées directement ici ===
csv_path = "deputes_et_contactes_tricouteuse.csv"
sheet_name = "mobibi_test"
credentials_path = (
    "credentials.json"  # Chemin vers votre fichier d'accès Google Cloud
)

# === 2. Lecture du fichier CSV ===
print(f"Lecture du fichier CSV : {csv_path}...")
df = pd.read_csv(csv_path)

# Remplacement des valeurs NaN par des chaînes vides pour éviter les erreurs
df = df.fillna("")

# === 3. Connexion à Google Sheets ===
print("Connexion à Google Sheets...")
gc = gspread.service_account(filename=credentials_path)

# === 4. Envoi des données ===
sh = gc.open(sheet_name)
worksheet = sh.get_worksheet(0)

print(f"Envoi des données vers le Google Sheet '{sheet_name}'...")
worksheet.clear()
worksheet.update([df.columns.values.tolist()] + df.values.tolist())

print("Importation réussie avec succès !")
```
